### Propiedades y pares de la SF en tiempo continuo

En la {ref}`tab:T3:SF:propiedades` se recogen las principales propiedades de la serie de Fourier para señales periódicas de período fundamental $T_0$ y potencia finita, mientras que en la {ref}`tab:T3:SF:pares` se muestran los pares transformados más comunes.

:::{table} Propiedades de la serie de Fourier
:name: tab:T3:SF:propiedades

| Propiedad | Señal periódica | Coef. Serie de Fourier |
| :--- | :--- | :--- |
|  | $\left.\begin{array}{c}\tilde{x}(t)\\ \tilde{y}(t)\end{array}\right\}$ Período $T_0$ <br> ($\omega_0=\frac{2\pi}{T_0}$) | $\begin{array}{c}c_k\\ d_k\end{array}$ |
| **Linealidad** | $A\tilde{x}(t)+B\tilde{y}(t)$ | $A\,c_k+B\,d_k$ |
| **Desplazamiento temporal** | $\tilde{x}(t-t_0)$ | $c_k\,e^{-jk\omega_0 t_0}$ |
| **Desplazamiento en frecuencia** | $e^{jM\omega_0 t}\tilde{x}(t)$ | $c_{k-M}$ |
| **Inversión temporal** | $\tilde{x}(-t)$ | $c_{-k}$ |
| **Conjugación** | $\tilde{x}^*(t)$ | $c^*_{-k}$ |
| **Escalado temporal** | $\tilde{x}(\alpha t)$, $\alpha>0$ <br> *(Periódica con período $T_0/\alpha$)* | $c_k$ |
| **Convolución periódica** | $\tilde{x}(t)\circledast \tilde{y}(t)$ | $T_0\,c_k d_k$ |
| **Multiplicación** | $\tilde{x}(t)\tilde{y}(t)$ | $c_k * d_k$ |
| **Diferenciación** | $\dfrac{d}{dt}\tilde{x}(t)$ | $jk\omega_0\,c_k$ |
| **Integración** | $\displaystyle \int_{-\infty}^{t}\tilde{x}(\tau)\,d\tau$ | $\left(\dfrac{1}{jk\omega_0}\right)c_k$ <br> *(Finita y periódica sólo si $c_0=0$)* |
:::

**Relación de Parseval:**
La potencia media de la señal periódica equivale a la suma de las potencias de sus armónicos:

```{math}
:enumerated: false
	\frac{1}{T_0}\int_{T_0}|\tilde{x}(t)|^2dt = \sum_{k=-\infty}^{+\infty}|c_k|^2
```

```{note} Sobre la relación de Parseval

Esta relación es consecuencia de que las exponenciales complejas relacionadas armónicamente forman una base ortogonal del espacio de las señales periódicas de periodo $T_0$.

Se puede ver como el teorema de Pitágoras aplicado a espacios de señales de dimensión infinita: mediante la serie de Fourier descomponemos la señal en componentes perpendiculares entre sí, formando una base.
```

:::{table} Pares Básicos de Series de Fourier
:name: tab:T3:SF:pares

| Señal periódica $\tilde{x}(t)$ | Coef. Serie de Fourier $c_k$ |
| :--- | :--- |
| $\displaystyle \sum_{k=-\infty}^{+\infty}c_k e^{jk\omega_0t}$ | $c_k$ |
| $e^{j\omega_0t}$ | $c_1=1$<br>$c_k=0,\quad k\neq 1$ |
| $\cos \omega_0 t$ | $c_1=c_{-1}=\frac{1}{2}$<br>$c_k=0, \quad k\neq \pm 1$ |
| $\sin \omega_0 t$ | $c_1=-c_{-1}=\frac{1}{2j}$<br>$c_k=0, \quad k\neq \pm 1$ |
| $1$ | $c_0=1$<br>$c_k=0, \quad k\neq 0$ |
| **Onda cuadrada periódica**<br>$\tilde{x}(t)=\begin{cases}1,&\|t\|<T_1\\0,& T_1<\|t\|\leq \frac{T}{2}\end{cases}$<br>$\tilde{x}(t+T)=\tilde{x}(t)$ | <br> $\displaystyle \frac{\sin(k\omega_0 T_1)}{k\pi} = \frac{\omega_0 T_1}{\pi}\text{sinc}\left(\frac{k\omega_0T_1}{\pi}\right)$ |
| **Tren de deltas**<br>$\displaystyle \sum_{n=-\infty}^{+\infty}\delta(t - nT)$ | <br> $\displaystyle c_k=\frac{1}{T},\quad \forall k$ |
:::

A continuación veremos con más detalle cada una de las propiedades y un ejemplo de aplicación de cada una de ellas.

#### Linealidad

:::{important .simple icon=false} Linealidad
Consideramos dos señales $\tilde{x}(t)$ e $\tilde{y}(t)$ periódicas con el mismo periodo $T_0$[^foot1].
```{math}
:enumerated: false
    \begin{split}
        \tilde{x}(t) \SF c_k\\
        \tilde{y}(t) \SF d_k
    \end{split}
```

[^foot1]: Recordemos que si las dos señales periódicas no tienen el mismo periodo fundamental, para que la señal $\tilde{z}(t)$ sea periódica, ambas señales tienen que tener un periodo en común, que será el mínimo común múltiplo del periodo fundamental de cada una de ellas:
    ```{math}
        T_0 = M T_{0x} = N T_{0y}, \quad M, N \in \mathbb{Z}.
    ```
    Además, habrá que hacer el desarrollo en serie de Fourier de ambas señales con el periodo común $T_0$.


```{math}
    \boxed{\tilde{z}(t)=A\tilde{x}(t)+B\tilde{y}(t) \SF a_k=A c_k + B d_k}
```
:::

:::{solution} ejemplo:T3:SF:propiedades:linealidad:ejemplo2
:class: dropdown

Como el desarrollo en serie de Fourier de $\tilde{x}(t)$ e $\tilde{y}(t)$ está hecho con distinto periodo, **no podemos aplicar la propiedad de linealidad**.

Para poder aplicar la propiedad de linealidad, en primer lugar debemos buscar si ambas señales tienen un periodo en común[^foot1]:

```{math}
T_0=M T_{0x} = N T_{0y}, \, N, M \in \mathbb{Z} \Rightarrow T_0 = 1\, (M=1, N=2).
```

Obtenemos el desarrollo en serie de Fourier para $\tilde{y}(t)$ con periodo $T_0=2$ identificando con la ecuación de síntesis:
```{math}
    \tilde{y}(t)=\sum_{k=-\infty}^\infty d_k e^{jk2\pi t} \quad \Rightarrow \quad
    d_k=\begin{cases}
        \frac{-1}{2j}, & k=-2,\\
        \frac{1}{2j}, & k=2,\\
        0, & k\neq \pm 2.
    \end{cases}
```

```{math}
    \begin{split}
        \tilde{z}(t) &=3\cos(2\pi t)-5j\sin(2\pi t) \SF a_k=3c_k-5j d_k \\
        & \omega_0=2\pi, T_0=1.
    \end{split}
```

```{math}
        a_k=\begin{cases}
            \frac{5}{2}, & k=-2,\\
            \frac{3}{2}, & k=-1,\\
            \frac{3}{2}, & k=1,\\
            -\frac{5}{2}, & k=2,\\
            0, & k\neq \{\pm 1, \pm 2\}.
        \end{cases}
```

```{math}
    \tilde{x}(t) \SF c_k\\
```

```{math}
    \boxed{\tilde{y}(t)=\tilde{x}(t-t_0) \SF d_k=c_k e^{-jk\omega_0 t_0}}
```

:::{exercise} Ejemplo 3
:label: ejemplo:T3:SF:propiedades:desp_t:ejemplo1

```{math}
    \begin{split}
        \tilde{x}(t) &=\cos(2\pi t) \SF \quad 
        c_k=\begin{cases}
            \frac{1}{2}, & k=\pm 1,\\
            0, & k\neq \pm 1.
        \end{cases}\\
        & \omega_0=2\pi, T_0=1.
    \end{split}
```
```{math}
    \tilde{y}(t)=\cos\left[2\pi\left(t-\frac{1}{4}\right)\right]=\tilde{x}\left(t-\frac{1}{4}\right) \SF ??
```

:::

:::{solution} ejemplo:T3:SF:propiedades:desp_t:ejemplo1
:class: dropdown

```{math}
    \begin{split}
        \tilde{y}(t) &=\cos\left[2\pi\left(t-\frac{1}{4}\right)\right]=\tilde{x}\left(t-\frac{1}{4}\right) \SF d_k = c_k e^{-jk2\pi\left(\frac{1}{4}\right)} \\
        & \omega_0=2\pi, T_0=1.
    \end{split}
```

```{math}
        d_k=\begin{cases}
            \frac{1}{2} e^{-j(-1) 2\pi \left(\frac{1}{4}\right)}=\frac{1}{2} e^{j\pi/2}=\frac{j}{2}, & k=-1,\\
            \frac{1}{2} e^{-j(1) 2\pi \left(\frac{1}{4}\right)}=\frac{1}{2} e^{-j\pi/2}=\frac{-j}{2}, & k=1,\\
            0, & k\neq \pm 1.
        \end{cases}
```

Vamos a comprobarlo identificando con la ecuación de síntesis directamente:
```{math}
    \begin{split}
        \tilde{y}(t) &=\cos\left[2\pi\left(t-\frac{1}{4}\right)\right]=\frac{e^{j2\pi\left(t-\frac{1}{4}\right)}+e^{-j2\pi\left(t-\frac{1}{4}\right)}}{2}= \sum_{k=-\infty}^\infty d_k e^{jk2\pi t}\\
            &=\frac{1}{2}e^{-j\frac{\pi}{2}}e^{j2\pi t}+\frac{1}{2}e^{j\frac{\pi}{2}}e^{-j2\pi t} = \frac{j}{2}e^{-j2\pi t}-\frac{j}{2}e^{j2\pi t}.
    \end{split}
```

```{math}
        &=\frac{1}{2}e^{j8\pi t}+\frac{1}{2}e^{j4\pi t}.
    \end{split}
```

#### Simetría y conjugación

:::{important .simple icon=false} Simetría y conjugación
Consideramos dos señales $\tilde{x}(t)$ de periodo $T_0$.
```{math}
:enumerated: false
    \tilde{x}(t) \SF c_k\\
```
```{math}
    \boxed{\tilde{y}(t)=\tilde{x}(-t) \SF a_k=c_{-k}}
```
```{math}
    \boxed{\tilde{z}(t)=\tilde{x}^\ast(t) \SF b_k=c^\ast_{-k}}
```
:::

:::{important} Consecuencias:
- Si $\tilde{x}(t)$ tiene simetría par[^foot2], los coeficientes de su serie de Fourier también la tiene: 

  $\tilde{x}(t)=\tilde{x}(-t) \SF c_k = c_{-k}$

- Si $\tilde{x}(t)$ tiene simetría impar[^foot2], los coeficientes de su serie de Fourier también la tiene: 

  $\tilde{x}(t)=-\tilde{x}(-t) \SF c_k = -c_{-k}$

[^foot2]: No necesariamente real. Si la señal es compleja, tanto su parte real como su parte imaginaria tienen simetría par o impar en cada caso.

- Si $\tilde{x}(t)$ es real, los coeficientes de su serie de Fourier son hermíticos[^foot3]:

  $\tilde{x}(t)=\tilde{x}^\ast(t) \SF c_k^\ast = c_{-k}\quad$ (visto en {ref}`sec:T3:SF:reales`)

[^foot3]: Al tener simetría hermítica, su parte real (módulo) es par y su parte imaginaria (fase) es impar.

- Si $\tilde{x}(t)$ es **real y par**, los coeficientes de su serie de Fourier son **reales y pares**.

- Si $\tilde{x}(t)$ es **real e impar**, los coeficientes de su serie de Fourier son **imaginarios puros e impares**.

:::

````python
import numpy as np
from bokeh.plotting import figure, show
from bokeh.layouts import column, row
from bokeh.models import ColumnDataSource, CustomJS, Slider, Span, Label
from bokeh.io import output_notebook

output_notebook()

# ==========================================
# 1. PARÁMETROS Y DATOS INICIALES
# ==========================================
# Colores y Estilo
COLOR_SIG = '#7E3258'   # Granate
COLOR_PHASE = '#2ca02c' # Verde

# Parámetros físicos (Idénticos a tu ejemplo)
FACTOR_ESPACIADO = 2.5
RADIO_PULSO = 1.6
WIDTH = RADIO_PULSO
distancia = FACTOR_ESPACIADO * WIDTH
T0 = distancia      
w0 = 2 * np.pi / T0 
RANGO_T = T0 * 1.8
K_MAX_SHOW = 8 

# --- A. Datos para el Dominio del Tiempo ---
t_vals = np.linspace(-RANGO_T, RANGO_T, 1000)

# Función auxiliar Python para el estado inicial
def tri_pulse(t, w=1.0):
    return np.maximum(0, 1 - np.abs(t)/w)

# Generar datos iniciales (td = 0)
y_vals = np.zeros_like(t_vals)
y_ghost = np.zeros_like(t_vals)

for k in range(-4, 5):
    center = k * T0
    pulse = tri_pulse(t_vals - center, w=WIDTH)
    y_vals += pulse
    y_ghost += pulse # El fantasma se queda estático en td=0

# Fuentes de datos (DataSources)
source_signal = ColumnDataSource(data=dict(t=t_vals, y=y_vals))
source_ghost = ColumnDataSource(data=dict(t=t_vals, y=y_ghost))
# Fuente para el marcador 'td' (posición y etiqueta)
source_marker = ColumnDataSource(data=dict(x=[0], y=[0], text=["td"]))

# --- B. Datos para Frecuencia ---
k_indices = np.arange(-K_MAX_SHOW, K_MAX_SHOW + 1)

# Magnitud (Sinc cuadrado) - ESTÁTICA
def sinc_sq_envelope_k(k_val, w0, w=1.0):
    omega = k_val * w0
    # Evitar división por cero
    omega = np.where(omega == 0, 1e-10, omega) 
    return (np.sin(omega*w/2) / (omega*w/2))**2

mag_ck = sinc_sq_envelope_k(k_indices, w0, w=WIDTH)
# Calculamos envolvente continua para dibujar la línea punteada de fondo
k_cont = np.linspace(-K_MAX_SHOW, K_MAX_SHOW, 200)
mag_cont = sinc_sq_envelope_k(k_cont, w0, w=WIDTH)

source_mag = ColumnDataSource(data=dict(k=k_indices, mag=mag_ck, zeros=np.zeros_like(k_indices)))
source_mag_env = ColumnDataSource(data=dict(k=k_cont, mag=mag_cont))

# Fase - DINÁMICA (Inicialmente 0)
phase_vals = np.zeros_like(k_indices, dtype=float)
source_phase = ColumnDataSource(data=dict(k=k_indices, phase=phase_vals, zeros=np.zeros_like(k_indices)))


# ==========================================
# 2. CREACIÓN DE GRÁFICOS
# ==========================================

# --- Gráfico 1: Dominio del Tiempo ---
p_time = figure(width=550, height=400, title="Desplazamiento temporal")
style_math_axes(p_time, x_range=(-RANGO_T, RANGO_T), y_range=(-0.2, 1.4), 
                xlabel="t", ylabel=r"$$\tilde{x}(t-t_d)$$")

# Fantasma (Referencia estática)
p_time.line('t', 'y', source=source_ghost, line_dash='dashed', color=COLOR_SIG, alpha=0.3, line_width=1.5)
# Señal actual (Se mueve)
p_time.line('t', 'y', source=source_signal, color=COLOR_SIG, line_width=2.5)

# Elementos del marcador td (marca en el eje y etiqueta)
p_time.segment(x0='x', y0=0, x1='x', y1=0.1, source=source_marker, color="black", line_width=2)
p_time.text(x='x', y=-0.1, text='text', source=source_marker, text_align='center', 
            text_baseline='middle', text_font_style='italic')


# --- Gráfico 2: Magnitud (Estático) ---
p_mag = figure(width=400, height=200, title="Magnitud (Invariante)")
style_math_axes(p_mag, x_range=(-K_MAX_SHOW-1, K_MAX_SHOW+1), y_range=(0, 1.2), 
                xlabel="k", ylabel=r"$$|d_k|$$")

# Envolvente punteada
p_mag.line('k', 'mag', source=source_mag_env, line_dash='dotted', color='gray', alpha=0.6)
# Tallos (Stems)
p_mag.segment(x0='k', y0='zeros', x1='k', y1='mag', source=source_mag, color=COLOR_SIG, line_width=2)
p_mag.circle('k', 'mag', source=source_mag, color=COLOR_SIG, size=6)
# Fórmula estática
p_mag.add_layout(Label(x=K_MAX_SHOW, y=1.0, text=r"$$|d_k| = |c_k|$$", 
                       text_align="right", text_color="#555555"))


# --- Gráfico 3: Fase (Dinámico) ---
p_phase = figure(width=400, height=200, title="Fase (Lineal)")
style_math_axes(p_phase, x_range=(-K_MAX_SHOW-1, K_MAX_SHOW+1), y_range=(-3.5, 3.5), 
                xlabel="k", ylabel=r"$$\angle d_k$$")

# Líneas guía pi y -pi
p_phase.line([-100, 100], [np.pi, np.pi], line_dash='dashed', color='gray', alpha=0.3)
p_phase.line([-100, 100], [-np.pi, -np.pi], line_dash='dashed', color='gray', alpha=0.3)

# Tallos dinámicos
p_phase.segment(x0='k', y0='zeros', x1='k', y1='phase', source=source_phase, color=COLOR_PHASE, line_width=2)
p_phase.circle('k', 'phase', source=source_phase, color=COLOR_PHASE, size=6)


# ==========================================
# 3. LÓGICA DE INTERACCIÓN (JAVASCRIPT)
# ==========================================
slider = Slider(start=0, end=T0, value=0, step=T0/100, title=r"$$t_d$$ (Desplazamiento)")

# Todo el recálculo ocurre en el cliente (JS) para máxima fluidez
callback = CustomJS(args=dict(source_sig=source_signal, 
                              source_marker=source_marker,
                              source_ph=source_phase, 
                              slider=slider,
                              T0=T0, w0=w0, WIDTH=WIDTH), code="""
    const td = slider.value;
    
    // --- 1. RECALCULAR SEÑAL TEMPORAL ---
    const t = source_sig.data['t'];
    const y = source_sig.data['y'];
    
    // Reiniciamos array a ceros
    y.fill(0);
    
    // Sumamos pulsos (rango k suficiente para cubrir la vista)
    for (let k = -5; k <= 5; k++) {
        let center = k * T0 + td;
        for (let i = 0; i < t.length; i++) {
            // Lógica tri_pulse: max(0, 1 - |t - center|/w)
            let val = 1 - Math.abs(t[i] - center) / WIDTH;
            if (val > 0) {
                y[i] += val;
            }
        }
    }
    source_sig.change.emit();

    // --- 2. MOVER MARCADOR td ---
    source_marker.data['x'][0] = td;
    source_marker.change.emit();

    // --- 3. RECALCULAR FASE ---
    const k_arr = source_ph.data['k'];
    const ph = source_ph.data['phase'];
    const two_pi = 2 * Math.PI;

    for (let i = 0; i < k_arr.length; i++) {
        let k = k_arr[i];
        
        // Fase lineal: -k * w0 * td
        let angle = -k * w0 * td;
        
        // WRAPPING manual (Módulo entre -pi y pi en JS)
        // JS '%' devuelve resto con signo, necesitamos corrección para negativos
        let wrapped = ((angle + Math.PI) % two_pi);
        if (wrapped < 0) wrapped += two_pi;
        wrapped -= Math.PI;
        
        ph[i] = wrapped;
    }
    source_ph.change.emit();
""")

slider.js_on_change('value', callback)

# ==========================================
# 4. LAYOUT
# ==========================================
layout = column(
    slider,
    row(p_time, column(p_mag, p_phase))
)

show(layout)
````
